```python
rainbow = {'red', 'orange', 'yellow', 'green',
'blue', 'indigo', 'violet'}
olympic_flag = {'red', 'green', 'yellow', 'blue',
'black'}
rainbow.difference(olympic_flag)
#{‘orange’, ‘indigo’, ‘violet’}
olympic_flag.difference(rainbow)
#{‘black’}
```
